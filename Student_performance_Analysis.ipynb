```python
# handel missing value
# At first fill marks with missing marks
df['Marks']=df['Marks'].fillna(df['Marks'].mean())
print(df['Marks'])
```

```python
# fill missing city with the common city
df['City']=df['City'].fillna(df['City'].mode()[0])
print(df['City'])
```

```python
# fill attendence with avg
df['Attendance']=df['Attendance'].fillna(df['Attendance'].mean())
print(df['Attendance'])
```

```python
# check wheather any missing value remain
print("---------missing value----------")
print(" ")
print(df.isnull().sum())
```

```python
# calculate the avg marks for each city
city_avg=df.groupby('City')['Marks'].mean()
print("---------avg marks for each city----------")
print(" ")
print(city_avg)
```

```python
# find avg marks by each subject
subject_avg=df.groupby('Subject')['Marks'].mean()
print("---------avg marks by each subject----------")
print(" ")
print(subject_avg)
```

```python
# count number of students in each city
city_student_count=df.groupby('City').Student_ID.count()
print("---------number of students in each city----------")
print(" ")
print(city_student_count)
```

```python
# find avg attendence by city
city_attendance=df.groupby('City')['Attendance'].mean()
print("---------avg attendence by city----------")
print(" ")
print(city_attendance)
```

```python
# filter student who score above 80
above_80=df[df['Marks']>80]
print("---------student who score above 80----------")
print(" ")
print(above_80)
#
```

```python
# filter student from kolkata
kolkata=df[df['City']=='Kolkata']
print("---------student from kolkata----------")
print(" ")
print(kolkata)
```

```python
# filter student below 75 attendence
below_75=df[df['Attendance']<75]
print("---------student below 75 attendence----------")
print(" ")
print(below_75)
#
```

```python
# sort students by marks
sorted_marks=df.sort_values(by='Marks')
print("---------students by marks----------")
print(" ")
print(sorted_marks)
```

```python
# create a pass or fail column
df['Pass_Fail'] = df['Marks'].apply(lambda x: 'Pass' if x >= 80 else 'Fail')
print(df['Pass_Fail'])
print(df)
```

```python
# count pass or fail student
pass_fail_count=df['Pass_Fail'].value_counts()
print("---------pass or fail student----------")
print(" ")
print(pass_fail_count)
```

```python
# find the highest performance city
highest_performance_city=city_avg.idxmax() # idmax khuje ber korlo  performance besi
print("---------highest performance city----------")
print(" ")
print(highest_performance_city)
```

```python
# find the best perfomance subject
best_performance_subject=subject_avg.idxmax()
print("---------best perfomance subject----------")
print(" ")
print(best_performance_subject)
```

```python
# find avg marks for each gender
gender_avg=df.groupby('Gender')['Marks'].mean()
print("---------avg marks for each gender----------")
print(" ")
print(gender_avg)
```

```python
# calculate the corelation between attendance and marks
correlation=df['Attendance'].corr(df['Marks'])
print("---------corelation between attendance and marks----------")
print(" ")
print(correlation)
```

```python
# calculate avg marks for each subject
subject_avg=df.groupby('Subject')['Marks'].mean()
print("---------avg marks for each subject----------")
print(" ")
print(subject_avg)
print("")
# create a bar chart
subject_avg.plot(kind='bar',title='Average Marks by Subject',xlabel='Subject',ylabel='Average Marks')
```

```python
# create a gender wise comparison chat
gender_avg.plot(kind='bar',title='Average Marks by Gender',xlabel='Gender',ylabel='Average Marks')
```

```python
# find students with both good marks and good attendenc
good_marks_attendance=df[(df['Marks']>80) & (df['Attendance']>75)]
print("---------students with both good marks and good attendence----------")
print(" ")
print(good_marks_attendance)
chart=good_marks_attendance.plot(kind='bar',title='Students with Good Marks and Attendance',xlabel='Student ID',ylabel='Marks')
#
```

```python
#remove duplicate rows
df=df.drop_duplicates()
print("---------duplicate row removed----------")
print(" ")
print(df)
```
