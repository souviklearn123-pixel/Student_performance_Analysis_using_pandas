```python
#--------------------student prformance data visualization-----------------------
# Import Pandas library
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset from Google Drive
df = pd.read_csv("/content/drive/MyDrive/student_performance_dataset.csv")

print(df.head())
```

```python
# Basic information of the dataset
print("shape of the dataset:",df.shape)
print("columns name:")
print(df.columns)
print(" ")
print("dataset information:")
df.info()
```

```python
# statistical summary of numerical columns
df.describe()
```

```python
# avg marks by gender
gender_avg = df.groupby('Gender')['Marks'].mean()
display(gender_avg)
print("----Check it for bar plot-----")
print("")
gender_avg.plot(kind='bar')
plt.title('Average Marks by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Marks')
plt.xticks(rotation=0)
plt.show()
```

```python
# avg marks by subject
gender_avg = df.groupby('Subject')['Marks'].mean()
display(gender_avg)
print("----Check it for bar plot-----")
print("")
gender_avg.plot(kind='bar',color="grey",edgecolor="black")
plt.title('Average Marks by Subject')
plt.xlabel('Subject')
plt.ylabel('Average Marks')
plt.xticks(rotation=45)
plt.show()
```

```python
# distribution of marks by HISTOGRAM
plt.figure(figsize=(8, 6))
plt.hist(df['Marks'], bins=20, color='skyblue', edgecolor='black') # divide 10 interval
plt.title('Distribution of Marks')
plt.xlabel('Marks')
plt.ylabel("Number of Students ")
```

```python
# relationship between attendence and marks
plt.figure(figsize=(8, 6))
plt.scatter(df['Attendance'], df['Marks'], color='green', alpha=0.7)
plt.title('Attendance vs Marks')
plt.xlabel('Attendance')
plt.ylabel('Marks')
```

```python
# marks distribution by gender  boxplot
import seaborn as sns
plt.figure(figsize=(8, 6))
sns.boxplot(x='Gender', y='Marks', data=df, palette='pastel')
plt.title('Marks Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Marks')

```

```python
#Number of students by city  countplot
plt.figure(figsize=(8, 6))
sns.countplot(x='City', data=df, palette='pastel')
plt.title('Number of Students by City')
plt.xlabel('City')
plt.ylabel('Number of Students')
plt.xticks(rotation=45)
plt.show()
```

```python
# age distribution histogram
plt.figure(figsize=(8, 6))
sns.histplot(df['Age'], bins=20, kde=True, color='skyblue', edgecolor='black')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Number of Students')
plt.show()
```

```python
# average marks by city
city_avg = df.groupby('City')['Marks'].mean()
plt.figure(figsize=(8, 6))
city_avg.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Average Marks by City')
plt.xlabel('City')
plt.ylabel('Average Marks')
plt.xticks(rotation=45)
plt.show()
```

```python
# Marks distribution by subject
plt.figure(figsize=(8, 6))
sns.boxplot(x='Subject', y='Marks', data=df, color="skyblue")
plt.title('Marks Distribution by Subject')
plt.xlabel('Subject')
plt.ylabel('Marks')
plt.xticks(rotation=45)
plt.show()
```

```python
# relationship between age and marks
plt.figure(figsize=(8, 6))
sns.scatterplot(x='Age', y='Marks', data=df, color='skyblue', alpha=0.7)
plt.title('Age vs Marks')
plt.xlabel('Age')
plt.ylabel('Marks')
plt.show()
```

```python
# Correlation heatmap

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))
correlation = df[['Age', 'Marks', 'Attendance']].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='Blues',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap')
plt.show()
```

```python
# distribution of stident attendence
plt.figure(figsize=(8, 6))
sns.histplot(df['Attendance'], bins=20, kde=True, color='skyblue', edgecolor='black')
plt.title('Attendance Distribution')
plt.xlabel('Attendance')
plt.ylabel("number of students")
plt.show()
```

```python
# avg marks by age
age_avg=df.groupby('Age')['Marks'].mean()
plt.figure(figsize=(8, 6))
age_avg.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Average Marks by Age')
plt.xlabel('Age')
plt.ylabel('Average Marks')
plt.show()
```

```python
# avg attendence by  subject
subject_avg=df.groupby('Subject')['Attendance'].mean()
subject_avg.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Average Attendance by Subject')
plt.xlabel('Subject')
plt.ylabel('Average Attendance')
plt.show()
```

```python
# number of students by gender
gender_count = df['Gender'].value_counts()
sns.countplot(x='Gender', data=df, palette='pastel')
plt.title('Number of Students by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Students')
plt.show()
```

```python
# marks vs attendence with trend line
plt.figure(figsize=(8, 6))
sns.regplot(x='Attendance', y='Marks', data=df,scatter_kws={'color':'skyblue','edgecolor':'black'}, line_kws={'color': 'red'})
plt.title('Marks vs Attendance with Trend Line')
plt.xlabel('Attendance')
plt.ylabel('Marks')
plt.show()
```

key Insights
* Age and Marks show a very weak negative correlation (-0.08).
* Age and Attendance show a weak positive correlation (0.24).
* Marks and Attendance have almost no correlation (0.03).
* Overall, the selected numerical variables show weak relationships with each other.
